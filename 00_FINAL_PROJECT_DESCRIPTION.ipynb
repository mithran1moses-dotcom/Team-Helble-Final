### Streamlit vs. a html file that is a web app

**It is 100% fine to have your website just be an HTML file inside of a GitHub repo, and not use streamlit at all.**

Claude Code and Gemini and Chat are all wonderful at designing web applications. I've built several this year. You would just be basically a "product manager" and instruct it on how to make changes.

I often suggest Streamlit because it's a package where the learning curve to making a website is a lot lower for the typical student in the past. Meaning, students can actually understand how it works.

Conversely, if you build an HTML web application you just won't understand anything inside of it and how it is made: but that might be TOTALLY okay. (And one advantage of an HTML file is that people still in a few years for sure, whereas with streamlit, you need to take a few safeguards to ensure the it continues to work as packages get upgrades.)

If you choose to not use Streamlit, you will still use GitHub to make the website - the repo just needs to have the HTML file plus any data files that it needs to have access to, and you set up a "Github Page" for the repo. It's a few clicks.

## TIPS

### GitHub, with Colab where it helps

You set up a team repo in Lecture 15 and shipped the WACC card through pull requests. **Keep working that way.** The project lives in your team's GitHub repo. Each person works in a branch, and changes come in through pull requests that a teammate reviews.

- **Colab still works** for notebooks: open a notebook from your repo in Colab, then save it back with *File → Save a copy in GitHub*. Streamlit apps and `.py` pipeline scripts are easier in VS Code with a coding agent.
- **Big raw data doesn't go in the repo.** Git-ignore it, and have your download code recreate it (or link to where it lives).
- **No API keys in the repo, ever.** See the Quick Reference, _API keys & secrets_.
- Merge conflicts happen. If you get stuck, commit what you have in your branch and ask a teammate (or me) before you force anything.

### Build your dashboard and your code with miniature data!

I think it's a good idea during development to work on a "small working version of the data".

This should look like your real data - try to make the variable names the same. This should have the same "unit" of observation. But you want it to be small enough that you can try to manually check all your work. Often, 5-10 firms worth of data and a subset of the sample period is enough. Usually, I only keep my variable of interest, identifying variables I'd need to merge with other datasets, and a few X variables.

This idea - to use small versions of data where you can hand check - is called "test units" in programming lingo. The "toy data" in assignment  4 and 5 are an example. You can use a small test version of the data for huge parts of your code, but you can also use it to verify subparts of your code too.


### Planning and file/folder structure

This is the perfect time to reiterate some wisdom to help make your lives easier on this project and beyond. The textbook chapter on [good analysis practices](https://ledatascifi.github.io/ledatascifi-2025/content/02/10_Golden_1.html) is golden here.

**I can NOT emphasize enough that this project is where you should put all the lessons on planning and pseudocoding into practice.**

1. Aim to break up your files into "purpose-specific" tasks.
    - For _example_: "01 Download returns and factors" > "02 Calculate Factor Loadings" > "03 Get Alphas" > "04 Build final analysis dataset(s)" > "05 Analysis and Report"
2. **To figure out what files you'll need, what each file does/what to name them... WORK BACKWORDS!**
    - Write out (on paper, it's always so much easier) exactly how your dataset needs to look (write out 5-10 fake rows of data with the variables you'll need).
    - Then figure out how to get from your raw data files to there. You might have to have separate processes/code steps for different variables. Again, I usually write this on paper or a whiteboard at home.
    - I've installed a big dang (8'x4') whiteboard in every apartment and office I've had since I was 20. It's revolutionary for work productivity. You can do so for under $20 at home depot/lowes. [Huge](https://www.lowes.com/pd/47-75-in-x-7-98-ft-Smooth-White-Wall-Panel/3015239) (they can cut this to any size) or [medium](https://www.lowes.com/pd/23-75-in-x-3-98-ft-Marker-Board-Markerboard-Wainscot-Wall-Panel/3042205)
3. Planning analysis: **Getting to good visualizations**
    - Once you know what your final dataset(s) will look like, you can think about how you might manipulate the data and what graphs might show interesting/relevant patterns in the data
    - **I STRONGLY suggest "imagineering" your graphs on paper**. I almost always draw what I want my graphs to look like (what is the x/y axis, how many lines/dots, if they should be different colors (hues), etc) and then often redraw them as I think of improvements. This makes building graphs SO much easier.
    - Then you can save your graphs/dashboard plans as a mockup using Excalidraw
