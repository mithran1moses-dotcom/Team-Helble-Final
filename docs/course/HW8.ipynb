## Part 2: The proposal

Add `PROPOSAL.md` to your repo, with your team name and every member's name at the top. Put images (mockups, charts) in the repo and link them from the file.

Write it as if you're asking for funding: polished formatting, clean editing, readable by someone outside this class. Use Markdown headings, tables, and images so it reads well on GitHub. **Read the two example proposals in the `Final Project` folder first.**

**Not a change log.** Current AI models like to write documents as a history of what was done ("first we tried X, then we fixed Y"). I don't care what you've done. Your reader has very limited time and needs to understand what you're doing, what you want to show, and how it can be done. Make the proposal self-contained and purpose-built: someone who has never seen your repo should get it from this document alone.

It must cover:

1. **The question**, in one sentence, and who would pay for the answer.
2. **The data.**
    - The unit of observation: what one row is, with its columns and a few example rows (real ones from v0)
    - Sample period and sample conditions (restrictions on time, industry, data availability)
    - Must-have vs. nice-to-have variables
    - Each raw input: the source, how you get it, and where it lives in the repo
3. **The method.** The pipeline steps (they should match the files in your repo), the analysis you'll run, and why it fits the question.
4. **Threats and checks.** The biggest ways an auditor could break your project (look-ahead, survivorship, LLM leakage, merges, p-hacking...), and how you handle each.
5. **Dashboard mockup.** A picture of what the finished dashboard will look like. Any kind of picture works: a sketch, [Excalidraw](https://excalidraw.com), a screenshot, or an image a leading AI model generates and edits for you.
